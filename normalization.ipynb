```python
import pandas as pd

df = pd.read_csv("Iris.csv")

df.head()
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()

print(df.isnull().sum())
```

```python
from sklearn.preprocessing import MinMaxScaler

numeric_cols = df.select_dtypes(include='number').columns.drop('Id')

scaler = MinMaxScaler()

df[numeric_cols] = scaler.fit_transform(df[numeric_cols])

df.head()
```

```python
df[numeric_cols].describe()
```

```python
df.to_csv("Normalized_Iris.csv", index=False)

print("Normalized dataset saved successfully!")
```
